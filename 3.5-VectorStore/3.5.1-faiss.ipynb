```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
loader = TextLoader("../3.2-DataIngestion/speech.txt")
documents = loader.load()

documents
```

```python
text_splitter = CharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=30
)

docs = text_splitter.split_documents(documents)
```

```python
len(docs)
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)
```

```python
db = FAISS.from_documents(docs, embeddings)
```

```python
db
```

```python
query = "What is this speech about?"

results = db.similarity_search(query)

results
```

```python
len(results)
results[0].page_content
```

```python
retriever = db.as_retriever()
retriever.invoke(query)
```

```python
# Perform similarity search and return both the matching documents and their similarity scores
docs_and_score = db.similarity_search_with_score(query)

docs_and_score
#lower score more similarity
```

```python
# Convert the query text into an embedding vector
embedding_vector = embeddings.embed_query(query)

embedding_vector
```

```python
# Search FAISS using the embedding vector
docs_score = db.similarity_search_by_vector(embedding_vector)

docs_score
```

```python
# Save the FAISS vector store locally
db.save_local("faiss_index")
```

```python
# Load the saved FAISS vector store
new_db = FAISS.load_local(
    "faiss_index",
    embeddings,
    allow_dangerous_deserialization=True
)
```

```python
# Search the loaded FAISS database using the same query
docs = new_db.similarity_search(query)

docs
```
