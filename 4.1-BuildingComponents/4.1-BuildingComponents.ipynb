```python
import os
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv())

# LangSmith Tracking
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGCHAIN_PROJECT")

print("Google API Key loaded:", os.getenv("GOOGLE_API_KEY") is not None)
print("LangSmith API Key loaded:", os.getenv("LANGCHAIN_API_KEY") is not None)
```

```python
#we are creating a chat model object so langchain can communicate with it
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)
print(llm)
```

```python
response=llm.invoke("who is anjali yadav") #invoke-send this input to the model and get the response
print(response.content) #print the actual content of the response
#here we are directly asking the question from the model
```

```python
from langchain_core.prompts import ChatPromptTemplate
#making reusable prompt template,giving system a set of instructions
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert AI Engineer. Provide clear and accurate answers."),
    ("user", "{input}")
]) #here user is a placeholder where actual question will go
   #system-instructions for the model

print(prompt)
```

```python
# Connect the prompt template to the Gemini model
#chain is used to connect componenets together
chain = prompt | llm

# Run the chain with our input
#here the input is first getting formatted by the prompt template
response = chain.invoke({
    "input": "Can you tell me about LangSmith?"
})

# Print the actual text response
print(response.content)
```

```python
from langchain_core.output_parsers import StrOutputParser
#gemini response contains metadata object etc parser helps us convert into a format we actually want to work with
# Create an output parser that converts the model response into a plain string
output_parser = StrOutputParser()

print(output_parser)
```

```python
chain = prompt | llm | output_parser
###StrOutputParser = AI response → plain text string
response = chain.invoke({
    "input": "Can you tell me about LangSmith?"
})

print(response)
```
